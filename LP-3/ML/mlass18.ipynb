```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
f = lambda x: (x+3)**2
df_ = lambda x: 2*(x+3)
```

```python
def gradient_descent(lr, x=2, steps=200, tol=1e-6):
    path = [x]
    for _ in range(steps):
        step = lr*df_(x); x = x - step; path.append(x)     # x_new = x_old - lr * f'(x_old)
        if abs(step) < tol: break
    return np.array(path)
```

```python
path = gradient_descent(0.1)
print(f"Local minimum at x = {path[-1]:.4f}, y = {f(path[-1]):.6f}, iterations = {len(path)-1}")
```

```python
xs = np.linspace(-8,4,200)
fig,ax = plt.subplots(1,2,figsize=(13,4))
ax[0].plot(xs,f(xs)); ax[0].plot(path,f(path),"ro--",markersize=3); ax[0].set_title("Descent path from x = 2")
for lr in [0.01,0.1,0.5,0.9]:
    p = gradient_descent(lr); ax[1].plot(f(p),label=f"lr={lr} ({len(p)-1} steps)"); print("lr",lr,"->",len(p)-1,"iterations, x =",round(p[-1],4))
ax[1].set_xlabel("Iteration"); ax[1].set_ylabel("f(x)"); ax[1].legend(); ax[1].set_title("Effect of learning rate")
plt.show()
```
