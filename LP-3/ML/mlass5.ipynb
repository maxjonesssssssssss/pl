```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
R = np.array([[5,3,0,1],[4,0,0,1],[1,1,0,5],[1,0,0,4],[0,1,5,4]], dtype=float)  # 0 = not rated
users, items = R.shape; K = 2
np.random.seed(0)
P = np.random.rand(users,K); Q = np.random.rand(items,K)
alpha, beta = 0.002, 0.02                      # learning rate, regularization
```

```python
for step in range(5000):                       # gradient descent on known ratings only
    for i in range(users):
        for j in range(items):
            if R[i][j] > 0:
                e = R[i][j] - P[i] @ Q[j]
                Pi = P[i].copy()
                P[i] += alpha*(2*e*Q[j] - beta*P[i])
                Q[j] += alpha*(2*e*Pi   - beta*Q[j])
```

```python
R_hat = P @ Q.T
print("Original:\n", R)
print("Predicted:\n", np.round(R_hat,2))
mask = R > 0
print("RMSE on known ratings:", np.sqrt(np.mean((R[mask]-R_hat[mask])**2)))
for i,j in zip(*np.where(R==0)):
    print(f"User {i+1}, Item {j+1} -> predicted rating {R_hat[i][j]:.2f}")
```
