```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.datasets import make_circles
from sklearn.decomposition import PCA, KernelPCA
```

```python
X, y = make_circles(n_samples=400, factor=0.3, noise=0.05, random_state=42)
print("Data shape:", X.shape)
Xp = PCA(n_components=2).fit_transform(X)
Xk = KernelPCA(n_components=2, kernel="rbf", gamma=10).fit_transform(X)
```

```python
fig,ax = plt.subplots(1,3,figsize=(15,4))
for a,d,t in zip(ax,[X,Xp,Xk],["Original (circles)","Normal PCA","Kernel PCA (rbf)"]):
    a.scatter(d[:,0],d[:,1],c=y,cmap="coolwarm"); a.set_title(t)
plt.show()
```
